```python
import random
secret_number=random.randint(1,100)
print("Guessing Game")
while True:
    guess=int(input("Enter a number: "))
    if guess < secret_number:
        print("Your guessing number is smaller than the number.\n")
    elif guess > secret_number:
        print("Your guessing number is higher than the number.\n")
    else:
        print("\nWell Done!")
        print("That is the right number.")
```
